9.	Design and train the neural network that computes the functionality of XOR function.

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
```

```python
y = np.array([
    [0],
    [1],
    [1],
    [0]
])
```

```python
model = Sequential()
model.add(Dense(4, input_dim=2, activation='relu'))  # Hidden layer
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(X, y, epochs=1000, verbose=0)
```

```python
loss, accuracy = model.evaluate(X, y)
print(f"\nFinal Accuracy: {accuracy:.3f}")
```

```python
predictions = model.predict(X)
print("\nXOR Predictions:")
for i in range(len(X)):
    print(f"Input: {X[i]} => Predicted: {predictions[i][0]:.3f} => Rounded: {round(predictions[i][0])}")
```
